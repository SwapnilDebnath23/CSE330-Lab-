```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial import Polynomial
```

```python
coeff = np.array([7, -5, 4, 0.0, -2, 3])
p = Polynomial(coeff)

print(p)
```

```python
p_three = p.deriv(3)
p_th = p_three(-1)

p_prime = p.deriv(1)
p_p = p_prime(-1)

res = 2*p_th - (p_p/3)
print(res)
```

```python
x = np.linspace(-5, 4, 10)
p_p = p(x)

p_two = p.deriv(2)
p_t = p_two(x)

plt.plot(x, p_p, 'g')
plt.plot(x, p_t, 'r')
plt.show()
```

```python
def l_k(data_x, k ,x):
  l = 1
  for j in range(len(data_x)):
    if(j!=k):
      if(j%2==0):
        l *= (x-data_x[j])/(data_x[k]-data_x[j])

  return l
```

```python
def get_poly_lagrange(data_x, data_y, x_arr):
  n = len(data_x)
  res = np.zeros(len(x_arr))

  for i in range(len(x_arr)):
    for k in range(n):
      res[i] += data_y[n-k-1]*l_k(data_x, k, x_arr[i])

  return res
```

```python
data_x = np.array([1, 3, 6, 10])
data_y = np.array([4, 9, 12, 20])
x_arr = np.array([0, 2, 5, 8])
p = get_poly_lagrange(data_x, data_y, x_arr)

print(p)
```

```python
def diff(f, x, h):
  return (f(x+h)-f(x))/h
```

```python
coeff = np.array([5, -3, 2, 1])
p = Polynomial(coeff)
p_prime = p.deriv(1)
p_p = p_prime(2)
er = 0.1
h=0.8
cnt = 0

while True:
  res = diff(p, 2, h)
  error = abs(p_p-res)

  if(error<er):
    break

  h = h/4
  cnt = cnt+1

print(f'Required h: {h}\nTotal numbers of reductions: {cnt}')
```
